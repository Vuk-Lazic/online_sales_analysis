## Step 1: Declare variables:

```python
num_a = 5
num_b = 10
```

## Step 2: Calculate aritmetic operations:

```python
sum = num_a + num_b
diff = num_a - num_b
prod = num_a * num_b
ration = num_a / num_b
```

## Step 3: Show the results:

```python
print(sum)
print(diff)
print(prod)
print(ration)
```

## Conclusion
The sum is 15, the diffrence is -5, the product is 50, and the quotient is 0.5.
