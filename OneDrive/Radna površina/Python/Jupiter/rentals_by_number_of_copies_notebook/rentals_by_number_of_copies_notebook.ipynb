## Step 3: Calculate total copies / times borrowed ratio
We are going to create new column (borrowings_per_copy) with *times borrowed* to *total copies* ratio:

```python
df['borrowings_per_copy'] = df['times_borrowed'] / df['total_copies']
```

## Step 4: Sort and print final results
We are going to sort the values, select fisrt 50 and print the final results.

```python
sorted_df = df.sort_values(by='borrowings_per_copy', ascending=False)
top_50 = sorted_df.head(50)
top_50[['title', 'times_borrowed', 'total_copies', 'borrowings_per_copy']]
```
