porvi korak

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

drugi korak

```python
df = pd.read_csv("books.csv")
```

treci korak

```python
df.head()
```

cetvrti korak
grupisanje podataka po sekcijama pa grupisanje knjiga koje postoje unutar svake sekcije i broj izdavanje knjiga koje su te knjige napravile unutar svake sekcije

```python
inventory_gap = df.groupby('section').agg({
    'title': 'count',
    'times_borrowed': 'sum'
})
```

peti korak
pravljenje nove kolone titles_to_borrow_ratio

```python
inventory_gap['titles_to_borrow_ratio'] = inventory_gap['times_borrowed'] / inventory_gap['title']
```

sesti korak 
sortiranje vrednosti

```python
inventory_gap.sort_values(by=['titles_to_borrow_ratio'], ascending=False, inplace=True)
```

sedmi korak
prikaz podataka

```python
plt.figure(figsize=(10, 7))
sns.barplot(data=inventory_gap, x="titles_to_borrow_ratio", y="section")

plt.title("section efficiency")
plt.xlabel("Titiles to borrow ratio")
plt.ylabel("Section")
plt.tight_layout()
plt.show()
```
