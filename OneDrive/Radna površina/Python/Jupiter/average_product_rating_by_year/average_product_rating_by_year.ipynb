```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

drugi korak: ucitavanje podataka

```python
df = pd.read_csv("online_store_ratings_through_years.csv")
```

treci korak: prikazivanje prhiv pet redova

```python
df.head()
```

cetvrti korak: pretvaranje u datetime

```python
df["date"] = pd.to_datetime(df["date"])
```

peti korak: Generisanje nove kolone 

```python
df["year"] = df["date"].dt.year
```

sesti korak: izracunavanje prosecnih ocena po godinama

```python
avg_rating_by_year = df.groupby("year")["rating"].mean().reset_index()
```

sedmi korak: prikazivanje dijagrama

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=avg_rating_by_year, x="year", y="rating", marker="o")
 
plt.title("Average Product Rating by Year")
plt.xlabel("Year")
plt.ylabel("Average Rating")
plt.ylim(1, 5)
plt.grid(True)
plt.tight_layout()
plt.show()
```
