## Step 1: Import required libraries:
We are going to use Pandas to load the data and Matplotlib and Seaborn to create the charts.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Step 2: Load the data se:
We will uise a dataset containing informations about books:

catalog_position,title,author,year_published,genre,section,total_copies,times_borrowed,last_borrowed_date,rating,ratings_count,price,language,page_count,dimensions_width,dimensions_thickness,dimensions_height,catalog_shelf,catalog_row,catalog_row_number


```python
df = pd.read_csv("bookss.csv")
```

## Step 3: Show some sample data:
Show first 5 rows.

```python
df.head()
```

## Step 4: Generate new column:
We will generate a new column containing the century in which each book was published. The values will be derived based on the year the book was released.


```python
df["published_century"] = ((df["year_published"] - 1) // 100 + 1).astype("Int64")
```

## Step 5: Calclate average rating by century:
We woll group the data by century of publication, and then calculate the average rating for each resulting group.

```python
avg_rating_by_century = (
    df.groupby("published_century")["rating"].mean().reset_index()
)
```

## Step 6: Show results:
 We will use our data to create a line plot that shows the final results.

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=avg_rating_by_century, x="published_century", y="rating", marker="o")
 
plt.title("Average Book Rating by Century")
plt.xlabel("Century of Publication")
plt.ylabel("Average Rating")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Conclusion: 

This analysis shows a clear trend in average book ratings across centuries.
Books from the 123th entury recived the highset average ratings, after which there is a gradul decline reaching its lowes point around the 19th century.
From the 20th century onward, ratings begin to rise again, suggesting a renewe d apprecition or shift in library preferences in modern times.

The results may reflect changes in literary styles, historical context, reader expectation, or the acailability of books from earlier periods that have stood the test of time.
